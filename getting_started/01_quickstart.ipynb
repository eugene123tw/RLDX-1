# 01 — Quickstart

Goal: in ~5 minutes, import `rldx`, load a pre-trained checkpoint,
feed a dummy observation through the policy, and read out an action.

This notebook requires a CUDA-capable GPU. The first run downloads
the checkpoint (~15 GB) into your HuggingFace cache.


## 1. Environment check


```python
import rldx
import torch

print('rldx    :', rldx.__version__)
print('torch   :', torch.__version__)
print('cuda?   :', torch.cuda.is_available())
if torch.cuda.is_available():
    print('gpu    :', torch.cuda.get_device_name(0))
```

## 2. Load a pre-trained checkpoint

`RLDXPolicy` is the main inference entry point. Under the hood it:

1. Downloads the checkpoint (if `model_path` is a HuggingFace repo id),
2. Loads the RLDX-1 model via `RLDX.from_pretrained` in bfloat16,
3. Loads the `RLDXProcessor` from the `processor/` subdirectory of the checkpoint,
4. Attaches the embodiment-specific modality config.


```python
from rldx.data.embodiment_tags import EmbodimentTag
from rldx.serving.policy import load_policy

# Default checkpoint for the quick-start inference call.
MODEL = 'RLWRLD/RLDX-1.1-FT-ROBOCASA'

policy = load_policy(
    EmbodimentTag.GENERAL_EMBODIMENT,
    MODEL,
    device='cuda:0',
)
print('policy:', type(policy).__name__)
print('modality keys:', list(policy.get_modality_config().keys()))
```

## 3. Build a dummy observation

For this quickstart we target the `robocasa_panda_omron` embodiment:
three 256x256 RGB views (`left_view`, `right_view`, `wrist_view`),
a small state vector, and a natural-language instruction.

`RLDXPolicy.get_action` takes the observation grouped by modality:
`obs['video'][name]`, `obs['state'][name]`, and one instruction list
per sample under `obs['language']`. Video arrays are `uint8` shaped `(B, T, H, W, 3)`; state arrays are
`float32` shaped `(B, T, D)`.


```python
import numpy as np

B, Tv, Ts, H, W = 1, 4, 1, 256, 256
rng = np.random.default_rng(0)

obs = {
    # Video — one frame per view.
    'video': {
        'left_view':  rng.integers(0, 255, (B, Tv, H, W, 3), dtype=np.uint8),
        'right_view': rng.integers(0, 255, (B, Tv, H, W, 3), dtype=np.uint8),
        'wrist_view': rng.integers(0, 255, (B, Tv, H, W, 3), dtype=np.uint8),
    },
    # State — one timestep, matching the robocasa state schema.
    'state': {
        'end_effector_position_relative': np.zeros((B, Ts, 3), dtype=np.float32),
        'end_effector_rotation_relative': np.zeros((B, Ts, 4), dtype=np.float32),
        'gripper_qpos':                   np.zeros((B, Ts, 2), dtype=np.float32),
        'base_position':                  np.zeros((B, Ts, 3), dtype=np.float32),
        'base_rotation':                  np.zeros((B, Ts, 4), dtype=np.float32),
    },
    # Language — one instruction list per batch item.
    'language': {'annotation.human.action.task_description': [['pick up the mug']]},
}
```

## 4. Run inference

`get_action` returns a dict keyed by the action modalities in the
checkpoint's modality config; each value has the model's action
horizon as its second dimension.


```python
actions, info = policy.get_action(obs)

print('action keys:', list(actions.keys()))
for k, v in actions.items():
    print(f'  {k:32s} shape={tuple(v.shape)} dtype={v.dtype}')
```
