# 04 — Inference Server

For closed-loop evaluation against a simulator or a real robot,
loading the model in-process usually does not fit the latency /
memory budget. RLDX ships a small ZeroMQ inference server that
holds the model in GPU memory and answers `get_action` requests
over a socket.

This notebook launches the server as a subprocess, connects a
client, sends one request, and shuts the server down.

The canonical docs are in [`../docs/inference_server.md`](../docs/inference_server.md).


## 1. Start the server

`rldx/eval/run_rldx_server.py` is the entry point. The key flags:

| Flag | Purpose |
|---|---|
| `--model-path` | Checkpoint directory or HF repo id. |
| `--embodiment-tag` | `EmbodimentTag` value matching the client's observation format. |
| `--host` / `--port` | Where to bind the ZMQ socket. Default `127.0.0.1:5555`. |
| `--denoising-timesteps` | Override the flow-matching denoising schedule (e.g. `0.0 0.1 0.3 0.6`). |
| `--deactivate-memory` | Load a memory-trained checkpoint as a memoryless model. |


```python
import os
import subprocess
import sys
import time
from pathlib import Path

REPO = Path.cwd().parent
MODEL = 'RLWRLD/RLDX-1.1-FT-ROBOCASA'
PORT = 20099

server_cmd = [
    'uv', 'run', 'python',
    str(REPO / 'rldx/eval/run_rldx_server.py'),
    '--model-path', MODEL,
    '--embodiment-tag', 'GENERAL_EMBODIMENT',
    '--host', '127.0.0.1',
    '--port', str(PORT),
]
print('>>>', ' '.join(server_cmd))
# Uncomment to actually start. The server takes ~30-60s to load the
# model on the first run (checkpoint download + bf16 load).
# server = subprocess.Popen(server_cmd, cwd=REPO)
# time.sleep(60)  # TODO: poll the server health endpoint instead
```

## 2. Connect a client

`rldx.serving.server.PolicyClient` is the matching ZMQ
client. Its `get_action(obs, options)` takes the same grouped
observation `RLDXPolicy.get_action` takes, serialises it with
msgpack over the socket, and returns a `(actions, info)` tuple.
Validation runs server-side.


```python
import numpy as np
from rldx.serving.server import PolicyClient

client = PolicyClient(host='127.0.0.1', port=PORT)

B, Tv, Ts, H, W = 1, 4, 1, 256, 256
rng = np.random.default_rng(0)
obs = {
    'video': {
        'left_view':  rng.integers(0, 255, (B, Tv, H, W, 3), dtype=np.uint8),
        'right_view': rng.integers(0, 255, (B, Tv, H, W, 3), dtype=np.uint8),
        'wrist_view': rng.integers(0, 255, (B, Tv, H, W, 3), dtype=np.uint8),
    },
    'state': {
        'end_effector_position_relative': np.zeros((B, Ts, 3), dtype=np.float32),
        'end_effector_rotation_relative': np.zeros((B, Ts, 4), dtype=np.float32),
        'gripper_qpos':                   np.zeros((B, Ts, 2), dtype=np.float32),
        'base_position':                  np.zeros((B, Ts, 3), dtype=np.float32),
        'base_rotation':                  np.zeros((B, Ts, 4), dtype=np.float32),
    },
    'language': {'annotation.human.action.task_description': [['pick up the mug']]},
}

# actions, info = client.get_action(obs)
# for k, v in actions.items():
#     print(f'  {k:32s} shape={v.shape} dtype={v.dtype}')
```

## 3. Simulator observation format

The server only speaks the RLDX format. The RLDX-1 sim envs emit
flat `video.<key>` / `state.<key>` observations, so the rollout
worker translates on its side with `rldx/eval/sim/rldx_format.py`.
`parse(env_name, modality_config, observation_keys, action_keys)`
picks the env family from the name, matches the policy's keys against
what the env renders (LIBERO's `state.x/y/z`, the GR-1 and RoboCasa
camera names) and fails before the first step if anything is missing.
The returned translation's `observation` and `env_action` then convert
every step.

A real robot client builds the grouped observation directly; see
`RLDXPolicy` in `rldx/serving/policy.py`.
